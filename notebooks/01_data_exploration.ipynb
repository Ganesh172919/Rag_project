# AARAG — Data Exploration

Explore the datasets used for training and evaluation.

```python
import json
import pandas as pd
from pathlib import Path
from collections import Counter

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load router training data
data_dir = Path('../data/processed')

if (data_dir / 'router_train.json').exists():
    with open(data_dir / 'router_train.json') as f:
        router_data = json.load(f)
    
    df = pd.DataFrame(router_data)
    print(f'Total samples: {len(df)}')
    print(f'\nLabel distribution:')
    print(df['label'].value_counts().sort_index())
    
    # Label names
    label_names = {0: 'No Retrieval', 1: 'Single-Step', 2: 'Multi-Step', 3: 'Graph-Global'}
    df['label_name'] = df['label'].map(label_names)
    
    # Plot distribution
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    df['label_name'].value_counts().plot(kind='bar', ax=axes[0], color='steelblue')
    axes[0].set_title('Query Complexity Distribution')
    axes[0].set_ylabel('Count')
    
    df['query'].str.len().hist(ax=axes[1], bins=30, color='coral')
    axes[1].set_title('Query Length Distribution')
    axes[1].set_xlabel('Characters')
    
    plt.tight_layout()
    plt.show()
else:
    print('Run data_preparation.py first to generate training data.')
```
